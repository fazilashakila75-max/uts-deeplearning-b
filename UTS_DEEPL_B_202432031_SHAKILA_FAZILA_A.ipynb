  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432031_ShakilaFazilaAdhar
import copy
import re
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432031_ShakilaFazilaAdhar
# 1. Memuat dataset CSV ulasan aplikasi DANA
df = pd.read_csv('review_dana_labelled.csv')

# 2. Membersihkan data kosong pada kolom teks dan label sentimen
df = df.dropna(subset=['content', 'sentimen'])

# Fungsi pembersihan teks (case folding, hapus tautan, tanda baca, angka, emoji)
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['cleaned_content'] = df['content'].apply(clean_text)
# Buang ulasan yang kosong setelah dibersihkan, lalu rapikan index agar sejajar dengan corpus
df = df[df['cleaned_content'] != ''].reset_index(drop=True)

# Mapping label teks menjadi angka
categories = ['NEGATIVE', 'NEUTRAL', 'POSITIVE']
label_mapping = {'NEGATIVE': 0, 'NEUTRAL': 1, 'POSITIVE': 2}
df['label'] = df['sentimen'].str.upper().str.strip().map(label_mapping)
df = df.dropna(subset=['label']).reset_index(drop=True)
df['label'] = df['label'].astype(int)

print('Jumlah baris & kolom setelah dibersihkan:', df.shape)
print('Distribusi label:')
print(df['sentimen'].value_counts())

# Tokenisasi kata (split berdasarkan spasi)
corpus = [text.split() for text in df['cleaned_content']]
y_data = df['label'].values

# Pembagian data: 80% (latih+validasi) dan 20% uji, lalu validasi 10% dari data latih
idx_all = np.arange(len(df))
idx_trainval, idx_test = train_test_split(idx_all, test_size=0.2, random_state=42, stratify=y_data)
idx_train, idx_val = train_test_split(idx_trainval, test_size=0.1, random_state=42, stratify=y_data[idx_trainval])
print('Data latih:', len(idx_train), '| validasi:', len(idx_val), '| uji:', len(idx_test))

# Vocabulary dibangun HANYA dari data latih agar tidak bocor ke data uji
vocab_size = 10000
word_counts = Counter(word for i in idx_train for word in corpus[i])
word_to_idx = {'<PAD>': 0, '<UNK>': 1}
for word, _ in word_counts.most_common(vocab_size - 2):
    word_to_idx[word] = len(word_to_idx)

# Konversi teks menjadi sekuens indeks dengan panjang tetap 50 (padding di DEPAN)
max_seq_len = 50
def text_to_sequence(text_words):
    seq = [word_to_idx.get(word, 1) for word in text_words][:max_seq_len]
    return [0] * (max_seq_len - len(seq)) + seq

X_data = np.array([text_to_sequence(doc) for doc in corpus])

# 3. Konversi ke Tensor PyTorch
X_train = torch.tensor(X_data[idx_train], dtype=torch.long)
y_train = torch.tensor(y_data[idx_train], dtype=torch.long)
X_val   = torch.tensor(X_data[idx_val], dtype=torch.long)
y_val   = torch.tensor(y_data[idx_val], dtype=torch.long)
X_test  = torch.tensor(X_data[idx_test], dtype=torch.long)
y_test  = torch.tensor(y_data[idx_test], dtype=torch.long)

batch_size = 128
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=batch_size, shuffle=True, drop_last=True)
val_loader   = DataLoader(TensorDataset(X_val, y_val), batch_size=batch_size, shuffle=False)
test_loader  = DataLoader(TensorDataset(X_test, y_test), batch_size=batch_size, shuffle=False)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432031_ShakilaFazilaAdhar
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim, dropout=0.3):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)
        self.lstm = nn.LSTM(embedding_dim, hidden_dim, batch_first=True)
        self.batchnorm = nn.BatchNorm1d(hidden_dim)
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_dim, output_dim)

    def forward(self, x):
        embedded = self.embedding(x)
        _, (hidden, cell) = self.lstm(embedded)
        last_hidden = hidden[-1]
        out = self.batchnorm(last_hidden)
        out = self.dropout(out)
        logits = self.fc(out)
        return logits
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432031_ShakilaFazilaAdhar
# 1. Hyperparameter
epochs = 15
learning_rate = 0.001
embedding_dim = 64
hidden_dim = 64
output_dim = 3
dropout_rate = 0.3

# 2. Model, loss function, optimizer
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = SequenceModel(len(word_to_idx), embedding_dim, hidden_dim, output_dim, dropout=dropout_rate).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. Riwayat loss dan akurasi
train_losses, val_losses = [], []
train_accs, val_accs = [], []
best_val_loss = float('inf')
best_state = None
best_epoch = 0

# 4. Training loop
for epoch in range(epochs):
    # ---- Fase training ----
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()
    train_loss = running_loss / total
    train_acc = correct / total

    # ---- Fase validasi ----
    model.eval()
    v_loss, v_correct, v_total = 0.0, 0, 0
    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            v_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            v_total += labels.size(0)
            v_correct += (predicted == labels).sum().item()
    val_loss = v_loss / v_total
    val_acc = v_correct / v_total

    train_losses.append(train_loss); train_accs.append(train_acc)
    val_losses.append(val_loss); val_accs.append(val_acc)

    # Simpan bobot terbaik berdasarkan validation loss
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_state = copy.deepcopy(model.state_dict())
        best_epoch = epoch + 1

    print(f'Epoch [{epoch+1}/{epochs}] - Train Loss: {train_loss:.4f} - Train Acc: {train_acc*100:.2f}% '
          f'| Val Loss: {val_loss:.4f} - Val Acc: {val_acc*100:.2f}%')

# Pakai bobot dengan validation loss terendah untuk evaluasi
model.load_state_dict(best_state)
print(f'\nBobot terbaik diambil dari epoch {best_epoch} (Val Loss: {best_val_loss:.4f})')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432031_ShakilaFazilaAdhar
# 1. Grafik riwayat Loss dan Accuracy (train vs validation) terhadap epoch
epoch_range = range(1, epochs + 1)
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].plot(epoch_range, train_losses, color='#114F6A', linewidth=2, marker='o', label='Training Loss')
axes[0].plot(epoch_range, val_losses, color='#1FA4B1', linewidth=2, marker='s', label='Validation Loss')
axes[0].set_title('Grafik Loss Selama Pelatihan Model LSTM')
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Loss')
axes[0].grid(True, linestyle='--', alpha=0.5); axes[0].legend()
axes[1].plot(epoch_range, train_accs, color='#114F6A', linewidth=2, marker='o', label='Training Accuracy')
axes[1].plot(epoch_range, val_accs, color='#1FA4B1', linewidth=2, marker='s', label='Validation Accuracy')
axes[1].set_title('Grafik Accuracy Selama Pelatihan Model LSTM')
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Accuracy')
axes[1].grid(True, linestyle='--', alpha=0.5); axes[1].legend()
plt.tight_layout()
plt.show()

# 2. Evaluasi pada data uji (prediksi, skor probabilitas, dan label aktual)
model.eval()
all_probs, all_preds, all_targets = [], [], []
with torch.no_grad():
    for inputs, labels in test_loader:
        outputs = model(inputs.to(device))
        probs = torch.softmax(outputs, dim=1)
        all_probs.append(probs.cpu().numpy())
        all_preds.extend(probs.argmax(dim=1).cpu().numpy())
        all_targets.extend(labels.numpy())
all_probs = np.vstack(all_probs)
all_preds = np.array(all_preds)
all_targets = np.array(all_targets)
confidence = all_probs.max(axis=1)

test_acc = accuracy_score(all_targets, all_preds)
print(f'Akurasi Data Uji: {test_acc*100:.2f}%\n')
print(classification_report(all_targets, all_preds, target_names=categories, digits=4))

# 3. Confusion matrix
cm = confusion_matrix(all_targets, all_preds)
plt.figure(figsize=(6, 5))
plt.imshow(cm, cmap='GnBu')
plt.colorbar()
plt.xticks(range(3), categories); plt.yticks(range(3), categories)
for i in range(3):
    for j in range(3):
        plt.text(j, i, cm[i, j], ha='center', va='center',
                 color='white' if cm[i, j] > cm.max() / 2 else 'black')
plt.title('Confusion Matrix Data Uji')
plt.xlabel('Prediksi'); plt.ylabel('Aktual')
plt.tight_layout()
plt.show()

# 4. Perbandingan distribusi sentimen aktual vs prediksi
target_counts = pd.Series(all_targets).value_counts().sort_index()
pred_counts = pd.Series(all_preds).value_counts().sort_index()
x = np.arange(len(categories))
width = 0.35
plt.figure(figsize=(10, 5))
plt.bar(x - width/2, [target_counts.get(i, 0) for i in range(3)], width, label='Aktual (Ground Truth)', color='#114F6A')
plt.bar(x + width/2, [pred_counts.get(i, 0) for i in range(3)], width, label='Prediksi LSTM', color='#1FA4B1')
plt.title('Perbandingan Distribusi Sentimen Aktual vs Prediksi Model LSTM')
plt.xlabel('Kategori Sentimen'); plt.ylabel('Jumlah Sampel Ulasan')
plt.xticks(x, categories)
plt.grid(True, linestyle='--', alpha=0.5); plt.legend()
plt.tight_layout()
plt.show()

# 5. Skor (confidence) hasil prediksi
mean_conf = [confidence[all_preds == i].mean() if (all_preds == i).any() else 0 for i in range(3)]
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].bar(categories, mean_conf, color=['#114F6A', '#7FB8C4', '#1FA4B1'])
axes[0].set_ylim(0, 1)
axes[0].set_title('Rata-rata Skor Prediksi per Kelas')
axes[0].set_ylabel('Skor (Probabilitas Softmax)')
for i, v in enumerate(mean_conf):
    axes[0].text(i, v + 0.01, f'{v:.3f}', ha='center')
correct_mask = all_preds == all_targets
axes[1].hist(confidence[correct_mask], bins=20, alpha=0.8, color='#1FA4B1', label='Prediksi Benar')
axes[1].hist(confidence[~correct_mask], bins=20, alpha=0.8, color='#114F6A', label='Prediksi Salah')
axes[1].set_title('Distribusi Skor Prediksi (Benar vs Salah)')
axes[1].set_xlabel('Skor Prediksi'); axes[1].set_ylabel('Jumlah Ulasan')
axes[1].legend()
plt.tight_layout()
plt.show()

# 6. Contoh hasil prediksi beserta skornya
sample = pd.DataFrame({
    'Ulasan': df.loc[idx_test, 'content'].values,
    'Aktual': [categories[i] for i in all_targets],
    'Prediksi': [categories[i] for i in all_preds],
    'Skor': confidence.round(4)
})
sample['Ulasan'] = sample['Ulasan'].astype(str).str.slice(0, 80)
print(sample.sample(10, random_state=42).to_string(index=False))
```
